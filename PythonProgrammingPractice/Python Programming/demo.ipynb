```python
def func1(a, /, b):
    # / is like a separator, however, anything before / must be positional-only arguments,
    # while anything afterward can be whatever (like a keyword argument for example)
    print(a+b)

func1(3, b=5)
```

```python
def func2(a, *, b):
    # * is like a separator, however, anything after * must be keyword arguments,
    # while anything before can be whatever (like a positional argument for example)
    print(a-b)

func2(3, b=2)
```

```python
def func3(a, /, b, *, c):
    print(a+b+c)

func3(3, 2, c=5)
```

```python
def func4(*args, **kwargs):
    # The first one is a tuple of positional arguments, while the second one is a dictionary of keyword arguments
    # They are also optional parameters
    # However, the keyword arguments must come after the arguments
    print(args)
    print(kwargs)

func4(10, 9, 8, 7, 6, a=1, b=2, c=3, d=4, e=5)
```
